# US iOS mobile-native idea scoring

Reproducible v4.0 score calculation for six desk-research assessments dated 2026-07-15. Inputs are judgment intervals tied to each idea's evidence ledger; they are not observed market metrics. A table is used instead of a chart because different archetypes have different weights and the interval precision is low.

```python
weights = {
    'consumer_utility': {'problem_demand': 25, 'monetization_economics': 15, 'competitive_wedge': 20, 'distribution_access': 25, 'repeat_value_retention': 5, 'timing': 10},
    'consumer_subscription': {'problem_demand': 20, 'monetization_economics': 20, 'competitive_wedge': 15, 'distribution_access': 20, 'repeat_value_retention': 20, 'timing': 5},
    'b2b_prosumer': {'problem_demand': 25, 'monetization_economics': 25, 'competitive_wedge': 15, 'distribution_access': 15, 'repeat_value_retention': 15, 'timing': 5},
}

ideas = [
    {'idea': 'Dog Rehab Capture', 'archetype': 'b2b_prosumer', 'dims': {'problem_demand': (3,4,4), 'monetization_economics': (2,3,4), 'competitive_wedge': (1,2,3), 'distribution_access': (2,3,4), 'repeat_value_retention': (3,4,4), 'timing': (3,3,4)}, 'expected': (46,65,77)},
    {'idea': 'Card Condition & Custody', 'archetype': 'b2b_prosumer', 'dims': {'problem_demand': (2,3,4), 'monetization_economics': (2,3,4), 'competitive_wedge': (1,2,3), 'distribution_access': (2,3,3), 'repeat_value_retention': (2,3,4), 'timing': (2,3,3)}, 'expected': (37,57,73)},
    {'idea': 'Tox Motion Passport', 'archetype': 'consumer_subscription', 'dims': {'problem_demand': (3,4,5), 'monetization_economics': (2,3,4), 'competitive_wedge': (1,2,3), 'distribution_access': (1,2,3), 'repeat_value_retention': (2,3,4), 'timing': (3,4,4)}, 'expected': (38,58,77)},
    {'idea': 'RefundLoop', 'archetype': 'consumer_subscription', 'dims': {'problem_demand': (3,4,4), 'monetization_economics': (1,2,3), 'competitive_wedge': (1,1,2), 'distribution_access': (1,2,2), 'repeat_value_retention': (2,3,3), 'timing': (3,4,4)}, 'expected': (34,51,58)},
    {'idea': 'Package Damage Claim Kit', 'archetype': 'consumer_utility', 'dims': {'problem_demand': (2,3,4), 'monetization_economics': (1,2,3), 'competitive_wedge': (2,3,4), 'distribution_access': (1,2,3), 'repeat_value_retention': (1,1,2), 'timing': (2,3,3)}, 'expected': (31,50,68)},
    {'idea': 'Trip Disruption Evidence', 'archetype': 'consumer_utility', 'dims': {'problem_demand': (2,3,4), 'monetization_economics': (1,2,3), 'competitive_wedge': (1,2,3), 'distribution_access': (2,2,3), 'repeat_value_retention': (1,2,3), 'timing': (2,3,4)}, 'expected': (32,47,67)},
]

def rounded_score(idea, bound_index):
    total = sum(weights[idea['archetype']][dimension] * interval[bound_index] / 5 for dimension, interval in idea['dims'].items())
    return int(total + 0.5)

for idea in ideas:
    idea['calculated'] = tuple(rounded_score(idea, idx) for idx in range(3))
    assert idea['calculated'] == idea['expected'], (idea['idea'], idea['calculated'], idea['expected'])
```

```python
print(f"{'Idea':31} {'Archetype':22} {'Low':>4} {'Base':>5} {'High':>5}")
print('-' * 73)
for idea in ideas:
    low, base, high = idea['calculated']
    print(f"{idea['idea']:31} {idea['archetype']:22} {low:>4} {base:>5} {high:>5}")
```

## Decision rule

All six assessments have clear fatal gates, overall evidence grade C, and intervals that remain below the attractive threshold on the low bound. The v4.0 desk decision is therefore `TESTE_DEGER` for each. Portfolio test priority is not a naked-score sort: it also considers archetype, regulation/privacy burden, access to a test channel, duration, and stopping-rule cost.
